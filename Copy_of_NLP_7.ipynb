```python
import nltk
nltk.download('treebank')
nltk.download('universal_tagset')

from nltk.corpus import treebank
from nltk.tag import hmm
```

```python
data = treebank.tagged_sents(tagset='universal')

print("Sample Dataset:")
print(data[:2])
```

```python
train_data = data[:3000]
test_data = data[3000:3100]

print("Training sentences:", len(train_data))
print("Testing sentences:", len(test_data))
```

```python
trainer = hmm.HiddenMarkovModelTrainer()

model = trainer.train_supervised(train_data)

print("HMM Model Trained Successfully")
```

```python
sentence = "The students are learning machine learning".split()

result = model.tag(sentence)

print("POS Tagging using HMM and Viterbi:")
for word, tag in result:
    print(word, ":", tag)
```

```python
accuracy = model.evaluate(test_data)

print("Model Accuracy:", accuracy)
```
